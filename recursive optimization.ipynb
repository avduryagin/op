```python
def get_tau(index):
    if index is None:
        return 0
    fun=solution.debit_functions[index]
    ex=fun.executor
    weight=solution.executors[ex].weight
    weight=1
    tau=fun.tau*weight
    next_fun=fun.cortege_previous
    tau_=get_tau(next_fun)
    return tau+tau_

def get_length(index):
    if index is None:
        return 0
    fun=solution.debit_functions[index]
    ex=fun.executor
    weight=solution.executors[ex].weight
    delta=fun.cortege_b-fun.cortege_a
    tau=delta*weight
    next_fun=fun.cortege_previous
    tau_=get_length(next_fun)
    return tau+tau_

def get_count(index):
    if index is None:
        return 0
    fun=solution.debit_functions[index]
    ex=fun.executor
    weight=solution.executors[ex].weight
    #tau=fun.tau*weight
    next_fun=fun.cortege_previous
    count=get_count(next_fun)
    return count+1

def get_vec(index,taumin_scale=0,taumax_scale=1,deltamin_scale=0,deltamax_scale=1):
    #38,400 - scaling constants
    if index is None:
        return []
    fun=solution.debit_functions[index]
    ex=fun.executor
    weight=solution.executors[ex].weight  
    tau=fun.tau
    delta=fun.cortege_b-fun.cortege_a
    next_fun=fun.cortege_previous
    va=[(tau-taumin_scale)/(taumax_scale-taumin_scale),(delta-deltamin_scale)/(deltamax_scale-deltamin_scale),weight]    
    L=get_vec(next_fun)
    L.insert(0,va)
    return L

def get_index(index):
    if index is None:
        return []
    fun=solution.debit_functions[index]
    next_fun=fun.cortege_previous       
    L=get_index(next_fun)
    L.insert(0,index)
    return L
```
